```python
import ollama

prompt = "what is the capital of thailand"

response = ollama.chat(model="llama3.2", messages=[
    # {"role":"system", "content": "You are female. Answer with นะคะ"},
    {"role": "user", "content":prompt}
])

response["message"]["content"]
```
